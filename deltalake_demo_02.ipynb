```python
from datetime import date

orders_schema = 'order_id INT, order_date DATE, order_customer_id INT, order_status STRING'

data = [
  (5, date(2020, 1, 1), 4, 'IN PGROGRESS'),
  (6, date(2020, 1, 1), 5, 'CLOSE')
]

df = spark.createDataFrame(data, schema=orders_schema)
```

```python
display(df)
```

```python
path = '/Volumes/deltalake_demo/default/ordersdemo1'
```

```python
df.write.format('delta').mode('append').save(path)
```
